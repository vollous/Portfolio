```python
# importing required modules
from pypdf import PdfReader
```

```python
# creating a pdf reader object
# https://www.scribd.com/document/96304856/Resumo-detalhado-maias
reader = PdfReader('ilide.info-resumo-detalhado-maias-pr_434844b91e401d23146ccb1e99dfdaba.pdf')

# printing number of pages in pdf file
print(len(reader.pages))

text = "".join([page.extract_text().replace("\uf0b7", " -").replace("o o", " --") for page in reader.pages])[31:]
```
